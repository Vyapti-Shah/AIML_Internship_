**CLASSIFICATION CODE**

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split #Imports a function to split data into training and testing sets
                                                     #This helps us test how well the model works on unseen data
from sklearn.linear_model import LogisticRegression #Imports the Logistic Regression algorithm
                                                    #This is used for classification problems (0 or 1, Yes/No, Pass/Fail)
from sklearn.metrics import accuracy_score, classification_report #accuracy_score → Measures how many predictions were correct
                                                                  #classification_report → Shows precision, recall, F1-score for each class
```

Creating the Dataset

```python
X = np.array([[1], [2], [3], [4], [5], [6], [7], [8]])
#X = Input features (study hours)
#Each value is inside [] because ML models expect 2D input
#Shape: (8 samples, 1 feature)
```

```python
y = np.array([0, 0, 0, 0, 1, 1, 1, 1])
#y = Output labels
#0 = Fail, 1 = Pass
#This is what the model will learn to predict
```

Spliting Data

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42
)
#Splits data into:
#X_train, y_train → Used to train the model
#X_test, y_test → Used to test the model
#test_size=0.25 → 25% data for testing
#random_state=42 → Makes results reproducible
```

Creating and Training Model

```python
model = LogisticRegression() #Creates a Logistic Regression model object
model.fit(X_train, y_train) #Trains the model where the model learns the relationship between: Study hours (X_train) and Pass/Fail (y_train)
```

Making Predictions

```python
y_pred = model.predict(X_test) #Model predicts pass/fail for the test data
```

Evaluating Model

```python
print("Accuracy:", accuracy_score(y_test, y_pred)) #Calculates how many predictions were correct Example: 0.8 = 80% accuracy
print(classification_report(y_test, y_pred)) #Shows detailed performance: Precision, Recall and F1-score
```

Predicting New Data

```python
new_hours = [[5]] #New students studied for 5 hours
prediction = model.predict(new_hours) #Model predicts pass/fail for this new student
print("Pass" if prediction[0] == 1 else "Fail") #Converts 1 or 0 into a human-readable result
```

```python
# Step 1: Import libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

```python
# Step 2: Create sample dataset
# Example: Hours studied vs Pass(1)/Fail(0)
data = {
    'Hours': [1, 2, 3, 4, 5, 6, 7, 8],
    'Pass':  [0, 0, 0, 0, 1, 1, 1, 1]
}

df = pd.DataFrame(data)
```

```python
# Step 3: Define features (X) and target (y)
X = df[['Hours']]
y = df['Pass']
```

```python
# Step 4: Split into training and testing data
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42
)
```

```python
# Step 5: Create Logistic Regression model
model = LogisticRegression()

# Step 6: Train the model
model.fit(X_train, y_train)
```

```python
# Step 7: Make predictions
y_pred = model.predict(X_test)

# Step 8: Check accuracy
print("Accuracy:", accuracy_score(y_test, y_pred))

# Step 9: Confusion matrix
print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred))

# Step 10: Detailed report
print("Classification Report:")
print(classification_report(y_test, y_pred))

# Step 11: Predict new value
hours = [[5]]
prediction = model.predict(hours)
print("Prediction for 5 hours:", prediction)

#Input (Hours studied) → Output (Pass = 1, Fail = 0)
#If someone studies 5 hours, will they pass or fail?"
#The model calculates probability using the logistic regression formula and then converts it into a class:
#     If probability ≥ 0.5 → output = 1 (Pass)
#     If probability < 0.5 → output = 0 (Fail)
```

```python
prob = model.predict_proba(hours)
print(prob)
#40% chance Fail, 60% chance Pass
```

**COST** **FUNCTION CODE**

```python
import numpy as np

# Sigmoid function
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

# Cost function
def compute_cost(X, y, weights, bias):

    m = len(y)

    # Linear equation
    z = np.dot(X, weights) + bias

    # Predicted probability
    h = sigmoid(z)

    # Cost calculation
    cost = (-1/m) * np.sum(
        y * np.log(h) + (1 - y) * np.log(1 - h)
    )

    return cost
```

```python
# Example usage
X = np.array([[1], [2], [3], [4]])
y = np.array([0, 0, 1, 1])

weights = np.array([0.5])
bias = 0.0

cost = compute_cost(X, y, weights, bias)

print("Cost:", cost)

#Lower cost = better model
#Higher cost = worse model
```

**GRADIENT DESCENT CODE**

```python
import numpy as np

# Step 1: Sigmoid function
def sigmoid(z):
    return 1 / (1 + np.exp(-z))


# Step 2: Cost function
def compute_cost(X, y, w, b):

    m = len(y)

    z = np.dot(X, w) + b
    h = sigmoid(z)

    cost = (-1/m) * np.sum(
        y * np.log(h) + (1-y) * np.log(1-h)
    )

    return cost


# Step 3: Gradient Descent function
def gradient_descent(X, y, w, b, learning_rate, iterations):

    m = len(y)

    for i in range(iterations):

        # prediction
        z = np.dot(X, w) + b
        h = sigmoid(z)

        # gradients
        dw = (1/m) * np.dot(X.T, (h - y))
        db = (1/m) * np.sum(h - y)

        # update weights and bias
        w = w - learning_rate * dw
        b = b - learning_rate * db

        # print cost every 100 iterations
        if i % 100 == 0:
            cost = compute_cost(X, y, w, b)
            print(f"Iteration {i}, Cost: {cost}")

    return w, b
```

```python
# Step 4: Example dataset
X = np.array([[1], [2], [3], [4], [5]])
y = np.array([0, 0, 0, 1, 1])

# Step 5: Initialize parameters
w = np.zeros(X.shape[1])
b = 0

# Step 6: Train using gradient descent
w, b = gradient_descent(
    X, y,
    w, b,
    learning_rate=0.1, #Small learning rate → slow learning
                       #Large learning rate → fast learning (but may overshoot)
    iterations=1000
)

print("Final weights:", w)
print("Final bias:", b)


# Step 7: Prediction function
def predict(X, w, b):

    z = np.dot(X, w) + b
    h = sigmoid(z)

    return (h >= 0.5).astype(int)


# Step 8: Test prediction
prediction = predict(np.array([[5]]), w, b)
print("Prediction for 5:", prediction)
```

Vectorized Cost Function

```python
def compute_cost(X, y, w, b):

    m = len(y)

    z = np.dot(X, w) + b
    h = sigmoid(z)

    cost = (-1/m) * np.sum(
        y*np.log(h) + (1-y)*np.log(1-h)
    )

    return cost
```

Vectorized Gradient Descent

```python
def gradient_descent(X, y, w, b, lr, iterations):

    m = len(y)

    for i in range(iterations):

        z = np.dot(X, w) + b
        h = sigmoid(z)

        dw = (1/m) * np.dot(X.T, (h - y))
        db = (1/m) * np.sum(h - y)

        w -= lr * dw
        b -= lr * db

    return w, b
```

**FEATURE SCALING**  (converting features into the same scale so that no feature dominates others)

```python
import numpy as np

X = np.array([20, 30, 40])

X_scaled = (X - X.min()) / (X.max() - X.min())

print(X_scaled)
```

```python
import numpy as np

X = np.array([20, 30, 40])

mean = np.mean(X)
std = np.std(X)

X_scaled = (X - mean) / std

print(X_scaled)
```

SQUARED ERROR COST FUNCTION (Mean Squared Error)

Simple Python Code

```python
import numpy as np

def squared_error_cost(y_actual, y_pred):

    m = len(y_actual)

    cost = (1/(2*m)) * np.sum((y_pred - y_actual)**2)

    return cost

# example
y_actual = np.array([1, 2, 3, 4])
y_pred = np.array([1.5, 2.5, 2.0, 5.0])

cost = squared_error_cost(y_actual, y_pred)

print("Cost:", cost)
```

Without vectorization (using loop)

```python
import numpy as np

def compute_cost(X, y, w, b):

    m = len(y)

    # prediction
    y_pred = np.dot(X, w) + b

    # cost
    cost = (1/(2*m)) * np.sum((y_pred - y)**2)

    return cost

# example
X = np.array([[1], [2], [3], [4]])
y = np.array([2, 4, 6, 8])

w = np.array([2])
b = 0

print("Cost:", compute_cost(X, y, w, b))
```
