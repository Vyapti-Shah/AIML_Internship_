```python
from ultralytics import YOLO

# Load a model
model = YOLO("yolo26n.pt")  # load a pretrained model (recommended for training)

# Train the model
results = model.train(data="VisDrone.yaml", epochs=10, imgsz=640, batch=16, mosaic=1.0) #Enable mosaic
```

Without Mosaic:<br>
-Poor detection of very small objects<br>
-Lower mAP

With Mosaic:<br>
-Better small object detection<br>
-Higher mAP<br>
-Better generalization

Mosaic:<br>
-Combines 4 images<br>
-Objects appear at different scales<br>
-Small objects become relatively larger<br>
-Model sees more object density per image
